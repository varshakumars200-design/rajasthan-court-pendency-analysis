```python
import pandas as pd
df=pd.read_csv("rajasthan.csv")
df.head(10)
```

```python
#Load + column names + strp
df.columns=["Sl.No.","District","Civil","Criminal","Total"]
df["District"]=df['District'].str.strip()
```

```python
#Total row alag karni hai + cleaning
state_total=df[df['District']=='Total']
df=df[df['District']!='Total'].copy()
df['District']=df['District'].replace({"Beawer":"Beawar"})
df=df.drop(columns="Sl.No.")
```

```python
#check
assert(df['Civil']+df['Criminal']==df['Total']).all()
assert df['Total'].sum()==state_total['Total'].iloc[0]
```

```python
#naye columns chahiye
df['Criminal_pct']=(df['Criminal']/df['Total']*100).round(1)
df["Civil_pct"]=(df["Civil"]/df["Total"]*100).round(1)
df["Share_of_State_pct"]=(df['Total']/df['Total'].sum()*100).round(2)
df['Rank']=df['Total'].rank(ascending=False).astype(int)
df['Pendency_band']=pd.qcut(df['Total'],3,labels=["Low","Medium","High"])
```

```python
df.sort_values("Rank").to_csv("cleaned_rajasthan.csv",index=False)
df.sort_values("Rank").head(10)
```

```python
out=df.sort_values("Rank").rename(columns={'Rank':"District_Rank"})
out.to_csv("rajasthan_for_sql.csv",index=False)
```
